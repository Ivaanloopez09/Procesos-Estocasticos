## Implementación de una función para simular el PAS

El PAS se define como $$X_t=\sum_s Z_s$$
donde $Z_s$ vale $1$ con probabilidad $p$ y vale $-1$ con probabilidad $1-p$

```python
def simular_PAS(p,n,reps=1):
  # Input:
  # p: Probabilidad del PAS. Número entre 0 y 1
  # n: Número de pasos del PAS. Número natural
  # reps: Número de trayectorias a simular. Número natural
  # Output:
  # output: Nos tiene que devolver una matriz que tenga dimensiones (reps,n)
  # que contenga las trayectorias de las repeticiones

  # Generamos una matriz inicial y auxiliar de tamaño (reps,n) con 1 y -1
  # generados aleatoriamente con probabilidad p y 1-p respectivamente
  matriz_inicial = np.random.choice([1, -1], size=(reps, n), p=[p, 1-p])


  # Matriz de la simulación del PAS (tamaño (reps,n)) obtenida sumando las
  # posiciones A[i,1:j] de la matriz inicial para cada posición B[i,j] de la
  # matriz final
  matriz_pas = np.cumsum(matriz_inicial, axis=1) #el axis=1 es para que lo haga
  # por filas

  return matriz_pas  # Matriz de tamaño (reps, n)
```

```python
simular_PAS(0.5,10,2)
```

## Visualización 2: diferentes trayectorias del PAS (con media y desviación típica teóricas)

Vamos a escribir código para crear las imágenes de las diapositivas 30 ($p=0.5$), 31 ($p=0.75$), 32 ($p=0.95$) del tema 1.

**Caso $p=0.5$**

```python
p=0.5 #probabilidad

# eje x
x = np.linspace(0, 500, 500)

# valores que se toman
y= simular_PAS(p,500,20)

media = x * (2*p - 1) #calculamos la media teórica del PAS
std = np.sqrt(4*p*(1-p)*x) #calculamos la desviación típica teórica

# banda de 2 desviaciones típicas
plt.fill_between(x, media - 2*std, media + 2*std, alpha=0.2,color='grey', label="±2σ") #El alpha es la transparencia del sombreado

# banda de 1 desviación típica
plt.fill_between(x, media - std, media + std, alpha=0.4,color='grey',label="±σ")

# dibujamos la media
plt.plot(x, media,'--',color='black', label='media teórica')

# trayectorias
plt.plot(x, y.T,linewidth=0.3)

#título
plt.title("20 trayectorias de un PAS con p=0.5", fontsize=10);

#leyenda
plt.legend(); # necesita los labels de antes

#cuadrícula
plt.grid(True, linestyle='--', alpha=0.6);
```

## Visualización 3: visualización de la media empírica del PAS cuando aumentamos el número de trayectorias (junto con la media teórica)

Vamos a escribir código para crear la imagen de la diapositiva 35.

Vamos a dibujar las medias empíricas de las trayectorias y vamos a ir aumentando el número de trayectorias, esperando ver como tiende entonces esta media hacia la media teórica.

```python
p=0.5 #probabilidad

# eje x
x = np.linspace(0, 500, 500)

#inicializamos un vector
v=[5,100,200,500,1000]

# para los colores
grises = np.linspace(0, 0.8, len(v))
# 0 = negro, 0.8 = gris claro (evitamos blanco puro)

# valores que se toman
for i, g in zip(v, grises): #el zip es para que i y g avancen sincronizados
    y= simular_PAS(p,500,i)
    mediaemp = np.mean(y, axis=0) #calculamos la media de nuestras trayectorias del PAS
    plt.plot(x,mediaemp,'-',linewidth=0.8, color=str(g),  # escala de grises
             label='n= %d' %i)

media = x * (2*p - 1) #calculamos la media teórica del PAS
std = np.sqrt(4*p*(1-p)*x) #calculamos la desviación típica teórica

# dibujamos la media teórica
plt.plot(x, media,'-',color='red', label='media exacta')

#título
plt.title("La media empírica tiende a la teórica", fontsize=10);

#leyenda
plt.legend(); # necesita los labels de antes

#cuadrícula
plt.grid(True, linestyle='--', alpha=0.6);
```
